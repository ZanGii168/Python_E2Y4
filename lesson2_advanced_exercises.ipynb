## Exercise 1: Numeric analysis

- Create one `int`, one `float` (scientific notation), and one `complex` number.
- Print each value, type, and (for complex) real/imag parts.
- Convert int->float and float->int, then print results.

```python
# Write your solution here
# Exercise 1: Numeric analysis

# 1. Create one int, one float (scientific notation), and one complex number
i = 42
f = 6.022e23
c = 3 + 4j

# 2. Print each value and its type (and real/imag parts for the complex)
print("int:    ", i, type(i))
print("float:  ", f, type(f))
print("complex:", c, type(c))
print("  real part:", c.real)
print("  imag part:", c.imag)

# 3. Convert int -> float and float -> int, then print results
i_to_f = float(i)
f_to_i = int(f)

print("\nint -> float:", i_to_f, type(i_to_f))
print("float -> int:", f_to_i, type(f_to_i))

# Note: int() truncates toward zero, it does not round
print("int(3.99) =", int(3.99))
```

## Exercise 2: Safe casting challenge

- Given `values = ["25", "3.14", "-8", "hello"]`, decide how each text should be converted.
- Use string methods (such as `isdigit()`, `replace()`, `lstrip()`) and `if/elif/else` to check each value.
- Convert to `int` if it is a whole number, to `float` if it is a decimal number, otherwise keep it as a string.
- Print each converted value together with its type.

```python
# Write your solution here
values = ["25", "3.14", "-8", "hello"]

for text in values:
    # Remove a single leading minus sign so negatives can be checked too
    if text.startswith("-"):
        body = text[1:]
    else:
        body = text

    if body.isdigit():
        # Only digits: whole number
        converted = int(text)
    elif body.replace(".", "", 1).isdigit():
        # Digits with exactly one decimal point: decimal number
        converted = float(text)
    else:
        # Anything else stays a string
        converted = text

    print(converted, type(converted))
```

## Exercise 3: String slicing matrix

- Use `text = "DataSciencePython"`.
- Print slices: first 4 chars, last 6 chars, every 2nd char, reversed text.
- Extract and print `Science` from the string using slicing only.

```python
# Write your solution here
text = "DataSciencePython"

print(text[:4])      # first 4 chars
print(text[-6:])     # last 6 chars
print(text[::2])     # every 2nd char
print(text[::-1])    # reversed text

print(text[4:11])    # Science
```

## Exercise 4: String transformation pipeline

- Start with `raw = "   python,java,c++ , go  "`.
- Strip the outer spaces and split the text by comma.
- Print the split list.
- Loop through the list, strip each language, uppercase it, and print it.

```python
# Write your solution here
raw = " python,java,c++ , go "

# Strip the outer spaces, then split by comma
languages = raw.strip().split(",")

# Print the split list
print(languages)

# Strip each language, uppercase it, and print it
for lang in languages:
    print(lang.strip().upper())
```

## Exercise 5: Formatting and escaping

- Create variables `product`, `price`, and `discount`.
- Print original and discounted price using f-string with 2 decimals.
- Print one line containing escaped quotes and another with newline+tab formatting.

```python
# Write your solution here
product = "Wireless Mouse"
price = 29.99
discount = 0.15  # 15% off

discounted_price = price * (1 - discount)
print(f"Product: {product}")
print(f"Original price: ${price:.2f}")
print(f"Discounted price ({discount:.0%} off): ${discounted_price:.2f}")

print("The product is called \"Wireless Mouse\" and it's on sale!")
print("Price details:\n\tOriginal: $29.99\n\tDiscounted: $25.49")
```
